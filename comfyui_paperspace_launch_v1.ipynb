NOTEBOOKIDとCLUSTERIDを適宜変更のうえブラウザで次を開きます。

- **NOTEBOOKID**: GradientコンソールのNotebook一覧/詳細で確認できるID
- **CLUSTERID**: デフォルトは `clg07azjl`（Private Clusterの場合はClustersページのID）


```python
%%bash
# 初回セットアップ（次回以降は必要なし）
set -euo pipefail

VENV="./venv"
PY="${VENV}/bin/python"
PIP="${PY} -m pip"

# venvが無ければ作る
if [ ! -x "${PY}" ]; then
  python3 -m venv "${VENV}"
fi

# ComfyUIが無ければ取得（既にあるなら更新）
if [ ! -d "/notebooks/ComfyUI/.git" ]; then
  git clone https://github.com/comfyanonymous/ComfyUI.git /notebooks/ComfyUI
else
  (cd /notebooks/ComfyUI && git pull --ff-only) || true
fi

# pip更新
${PIP} install --upgrade pip

# PyTorch（cu121）: 既存手順を踏襲（必要ならバージョンだけここで管理）
${PIP} install torch==2.4.0 torchvision==0.19.0 torchaudio==2.4.0 --index-url https://download.pytorch.org/whl/cu121

# ComfyUI本体＋カスタムノード依存
if [ -f "/notebooks/ComfyUI/requirements.txt" ]; then
  ${PIP} install -r /notebooks/ComfyUI/requirements.txt
fi

if [ -f "/notebooks/ComfyUI/custom_nodes/comfyui-manager/requirements.txt" ]; then
  ${PIP} install -r /notebooks/ComfyUI/custom_nodes/comfyui-manager/requirements.txt
fi

if [ -f "/notebooks/ComfyUI/custom_nodes/comfyui-impact-pack/requirements.txt" ]; then
  ${PIP} install -r /notebooks/ComfyUI/custom_nodes/comfyui-impact-pack/requirements.txt
fi

echo "Setup done: ${PY}"
```

```python
%%bash
set -euo pipefail

VENV="./venv"
PY="${VENV}/bin/python"

if [ ! -x "${PY}" ]; then
  echo "ERROR: venvがありません。先にセル1（初回セットアップ）を実行してください。" >&2
  exit 1
fi

# venvの site-packages を基準にLD_LIBRARY_PATHを組む（/usr/local 固定をやめる）
SITE="$(${PY} -c 'import site; print(site.getsitepackages()[0])')"

# torch/lib と nvidia/* の共有ライブラリ（存在しない場合もあるので、パス自体は入れてOK）
export LD_LIBRARY_PATH="${SITE}/torch/lib:${SITE}/nvidia/nvjitlink/lib:${SITE}/nvidia/cusparse/lib${LD_LIBRARY_PATH:+:${LD_LIBRARY_PATH}}"

# スレッド/BLAS系（必要最小限）
export MKL_THREADING_LAYER=GNU
export OMP_NUM_THREADS=1

# headless 安定化
export MPLBACKEND=Agg

# ---- 起動 ----
${PY} /notebooks/ComfyUI/main.py \
  --listen 0.0.0.0 --port 6006 --enable-cors-header \
  --reserve-vram 2 \
  --preview-method auto \
```
